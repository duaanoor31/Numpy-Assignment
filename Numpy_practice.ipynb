```python
import numpy as np
```

```python
# 2 Python List VS Numpy Array:
import numpy as np
marks = np.array([70, 80, 90, 60])
print(marks)
```

```python
#3 Creating Numpy Array:
#one dimensional array:
numbers = np.array([10, 20, 30, 40])
print(numbers)
#Two dimensional array:
marks = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [88, 92, 95]
])

print(marks)
```

```python
# 4 important array properties:
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(data.ndim)
print(data.shape)
print(data.size)
print(data.dtype)
```

```python
# 5 Indenxing:
numbers = np.array([10, 20, 30, 40, 50])
print(numbers[3])
print(numbers[-1])
```

```python
# 6 Indexing in a 2D Array:
marks = np.array([
    [70, 80, 90],
    [60, 75, 85]
])
print(marks[0, 2])
```

```python
# 7 Slicing
numbers = np.array([10, 20, 30, 40, 50])
print(numbers[0:4])
```

```python
# 8 vectorized operation:
numbers = np.array([10, 20, 30, 40])
print(numbers + 5)
print(numbers * 2)
print(numbers - 6)
print(numbers / 4)
```

```python
# 9 operation between two array:
a = np.array([10, 40, 30])
b = np.array([1, 2, 3])
print(a + b)
print(a * b)
```

```python
# 10 Mathematical Functions
numbers = np.array([4, 9, 16, 25])
print(np.sqrt(numbers))
numbers = np.array([2, 5, 4])
print(np.power(numbers, 2))
numbers = np.array([-10, -5, -7, 8])
print(np.abs(numbers))
numbers = np.array([4.2, 5.7, 8.9])
print(np.round(numbers))
```

```python
# 11 Aggregation function:
marks = np.array([70, 85, 60, 95, 80])
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.median(marks))
print(np.std(marks))
```

```python
#12 Condition in numpy:
marks = np.array([30, 50, 80, 45, 90, 20])
print(marks >= 50)
```

```python
#13 Filtering Data:
marks = np.array([30, 55, 82, 98, 90, 30])
passed = marks[marks >= 50]
print(passed)
failed = marks[marks < 50]
print(failed)
```

```python
# 14 Reshape:
numbers = np.array([1, 2, 3, 4, 5, 6])
new_array = numbers.reshape(2, 3)
print(new_array)
```

```python
#15 Flatten:
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(data.flatten())
```

```python
#16 usefull numpy function:
print(np.zeros(10))
print(np.ones(6))
print(np.arange(1, 21))
print(np.arange(0, 21, 2))
```

```python
#17 Random Numbers:
marks = np.random.randint(40, 100, 10)
print(marks)
```
